# Task 3
## Cleaning Data
#### Objectives: 
To Demonstrate professional-level data cleaning skills by taking a deliberately messy dataset and systematically transforming it into a clean, analysis-ready dataset. Document every decision. 
#### Dataset : 
I choosed "dirty_cafe_sales.csv" a Cafe sales dataset from kaggle.
#### Tech Stack:
Python, NumPy, Pandas, Jupyter Notebook.

```python
# importing libraries
import pandas as pd 
import numpy as np
df= pd.read_csv("dirty_cafe_sales.csv")
df.head()
```

```python
print("Dataset information:\n")
df.info()
```

```python
#Duplicate Removal
df_cleaned = df.drop_duplicates()
df_cleaned = df_cleaned.drop_duplicates(subset=["Transaction ID"]).copy()
df_cleaned
```

```python
#Standardize Missing Placeholders and Formatting
df_cleaned = df_cleaned.replace(
    ["ERROR", "UNKNOWN", "nan", "NaN", "null", "None", ""], np.nan
)
df_cleaned
```

```python
for text_col in ["Item", "Payment Method", "Location"]:
    df_cleaned[text_col] = df_cleaned[text_col].astype(str).str.strip().str.title()
    df_cleaned[text_col] = df_cleaned[text_col].replace("Nan", np.nan)
```

```python
# Standardize Transaction Date
df_cleaned["Transaction Date"] = pd.to_datetime(
    df_cleaned["Transaction Date"], errors="coerce"
)
df_cleaned["Transaction Date"] = (
    df_cleaned["Transaction Date"].ffill().bfill()
)
df_cleaned
```

```python
# Missing Data Imputation & Type Conversion
df_cleaned["Quantity"] = pd.to_numeric(
    df_cleaned["Quantity"], errors="coerce"
)
df_cleaned["Price Per Unit"] = pd.to_numeric(
    df_cleaned["Price Per Unit"], errors="coerce"
)
df_cleaned["Total Spent"] = pd.to_numeric(
    df_cleaned["Total Spent"], errors="coerce"
)
df_cleaned
```

```python
# Price Per Unit Imputation
df_cleaned["Price Per Unit"] = df_cleaned["Price Per Unit"].fillna(
    df_cleaned["Total Spent"] / df_cleaned["Quantity"]
)
item_price_map = {"Cookie": 1.0, "Tea": 1.5, "Coffee": 2.0, "Salad": 5.0}
df_cleaned["Price Per Unit"] = df_cleaned["Price Per Unit"].fillna(
    df_cleaned["Item"].map(item_price_map)
)
df_cleaned
```

```python
# Quantity Imputation
df_cleaned["Quantity"] = df_cleaned["Quantity"].fillna(
    df_cleaned["Total Spent"] / df_cleaned["Price Per Unit"]
)
df_cleaned["Quantity"] = df_cleaned["Quantity"].fillna(
    df_cleaned["Quantity"].median()
)
df_cleaned
```

```python
# Total Spent Calculation
df_cleaned["Total Spent"] = df_cleaned["Total Spent"].fillna(
    df_cleaned["Quantity"] * df_cleaned["Price Per Unit"]
)
```

```python
# Item Imputation
price_item_unique = {1.0: "Cookie", 1.5: "Tea", 2.0: "Coffee", 5.0: "Salad"}
df_cleaned["Item"] = df_cleaned["Item"].fillna(
    df_cleaned["Price Per Unit"].map(price_item_unique)
)
mode_p3 = df_cleaned[df_cleaned["Price Per Unit"] == 3.0]["Item"].mode()[0]
mode_p4 = df_cleaned[df_cleaned["Price Per Unit"] == 4.0]["Item"].mode()[0]
df_cleaned.loc[
    df_cleaned["Item"].isna() & (df_cleaned["Price Per Unit"] == 3.0), "Item"
] = mode_p3
df_cleaned.loc[
    df_cleaned["Item"].isna() & (df_cleaned["Price Per Unit"] == 4.0), "Item"
] = mode_p4
df_cleaned["Item"] = df_cleaned["Item"].fillna(df_cleaned["Item"].mode()[0])
```

```python
# Payment Method & Location Imputation
df_cleaned["Payment Method"] = df_cleaned["Payment Method"].fillna(
    df_cleaned["Payment Method"].mode()[0]
)
df_cleaned["Location"] = df_cleaned["Location"].fillna(
    df_cleaned["Location"].mode()[0]
)
```

```python
# Set Strict Data Types
df_cleaned["Transaction ID"] = df_cleaned["Transaction ID"].astype("string")
df_cleaned["Item"] = df_cleaned["Item"].astype("string")
df_cleaned["Quantity"] = df_cleaned["Quantity"].astype(int)
df_cleaned["Price Per Unit"] = df_cleaned["Price Per Unit"].astype(float)
df_cleaned["Total Spent"] = df_cleaned["Total Spent"].astype(float)
df_cleaned["Payment Method"] = df_cleaned["Payment Method"].astype("string")
df_cleaned["Location"] = df_cleaned["Location"].astype("string")
```

```python
# Save Cleaned Dataset to CSV
df_cleaned.to_csv("cleaned_cafe_sales.csv", index=False)
```
